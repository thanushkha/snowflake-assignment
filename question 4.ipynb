```
USE DATABASE TUTORIAL_DB;
USE SCHEMA SALES_SCHEMA;
USE WAREHOUSE COMPUTE_WH;
```

```
CREATE OR REPLACE TABLE TIME_TRAVEL_DEMO (
    ID INTEGER,
    NAME VARCHAR(50),
    CITY VARCHAR(50)
);
```

```
INSERT INTO TIME_TRAVEL_DEMO VALUES
(1, 'Ravi', 'Chennai'),
(2, 'Anu', 'Bangalore'),
(3, 'Kiran', 'Hyderabad'),
(4, 'Sneha', 'Tirupati');

SELECT * FROM TIME_TRAVEL_DEMO;
```

```
SET BEFORE_CHANGES = CURRENT_TIMESTAMP();
```

```
UPDATE TIME_TRAVEL_DEMO
SET CITY = 'Mumbai'
WHERE ID = 1;
```

```
DELETE FROM TIME_TRAVEL_DEMO
WHERE ID = 4;
```

```
SELECT * FROM TIME_TRAVEL_DEMO;
```

```
SELECT *
FROM TIME_TRAVEL_DEMO
AT (TIMESTAMP => $BEFORE_CHANGES);
```
